```python
from vllm import LLM, SamplingParams

def main():
    print("Loading model...")

    llm = LLM(
        model="facebook/opt-125m",
        enforce_eager=True,          # safer for debugging
        gpu_memory_utilization=0.8,  # fine for a tiny model
    )

    print("Generating...")

    sampling_params = SamplingParams(
        max_tokens=16,
        temperature=0.0,
    )

    outputs = llm.generate(
        ["chicken jockey"],
        sampling_params,
    )

    print("Output:")
    print(outputs[0].outputs[0].text)

if __name__ == "__main__":
    main()
```
